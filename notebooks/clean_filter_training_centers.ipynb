```python
import pandas as pd

# Load the CSV
file = "../data/raw/skill_india_training_centres.csv"

df = pd.read_csv(file, dtype=str).fillna("")

print("Before sorting:", len(df))
```

```python
# Sort:
# 1. State alphabetically
# 2. Centre name alphabetically within each state
df = df.sort_values(
    by=["state", "centre_name"],
    key=lambda x: x.str.strip().str.lower(),
    kind="stable"
).reset_index(drop=True)
```

```python
# Show every unique state exactly as it appears in the CSV
unique_states = sorted(df["state"].dropna().unique())

for state in unique_states:
    print(repr(state))
```

```python
# States/UTs to remove(not included under pm-ajay)
remove_states = [
   "andaman and nicobar islands",
    "dadra and nagar haveli",
    "daman and diu",
    "ladakh",
    "lakshadweep",
    "arunachal pradesh",
    "nagaland",
    "meghalaya",
    "mizoram"
]

# Remove matching states (case-insensitive)
df = df[
    ~df["state"].str.strip().str.lower().isin(
        [s.lower() for s in remove_states]
    )
].reset_index(drop=True)

# Save
df.to_csv(
    file,
    index=False,
    encoding="utf-8-sig"
)

print("Remaining records:", len(df))
print("Removed states:", remove_states)
```

```python
# Save back to the same CSV
df.to_csv(
    file,
    index=False,
    encoding="utf-8-sig"
)

print("Sorting completed.")
print("After sorting:", len(df))
```
